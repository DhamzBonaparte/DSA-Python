```python
class Customer:
  def __init__(self,name,age,gender,address):
    self.name=name
    self.age=age
    self.gender=gender
    self.address=address
  
  def print_address(self):
    print(self.address._Address__city,self.address.pin,self.address.state)
  
  def edit_profile(self,n_name,n_city,n_pin,n_state):
    self.name=n_name
    self.address.edit_address(n_city,n_pin,n_state)

class Address:
  def __init__(self,city,pin,state):
    self.__city=city
    self.pin=pin
    self.state=state
  
  def get_city(self):
    return self.__city
  
  def edit_address(self,n_city,n_pin,n_state):
    self.__city=n_city
    self.pin=n_pin
    self.state=n_state
```
