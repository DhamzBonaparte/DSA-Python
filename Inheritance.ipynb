```python
class Parent:
  def __init__(self,name,age,gender):
    self.name=name
    self.age=age
    self.gender=gender

  def show(self):
    return 'Parent: ' + self.name

class Children(Parent):
  def __init__(self,name,age,gender,address):
    super().__init__(name,age,gender)
    self.address=address
  
  def show2(self):
    return 'Child: ' + self.name + ', Age: ' + str(self.age) + ', Gender: ' + self.gender + ', Address: ' + self.address
```

```python
ch=Children('Sulav',22,'male','bkt')
ch.show()
ch.show2()
```
