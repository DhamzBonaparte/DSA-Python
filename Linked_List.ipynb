```python
class Node():
  def __init__(self,value):
    self.data=value;
    self.next=None
```

```python
class LinkedList():
  def __init__(self):
    self.head=None
    self.n=0

  def __len__(self):
    return self.n

  def insert_head(self,val):
    new_node=Node(val);
    new_node.next=self.head
    self.head=new_node
    self.n+=1

  def insert_mid(self,pos,val):

    if pos < 0 or pos > self.n:
      print("Index Error")

    new_node=Node(val)

    if pos==0:
      new_node.next=self.head
      self.head=new_node
      self.n+=1
      return

    curr=self.head
    for i in range(pos-1):
      curr=curr.next
    new_node.next=curr.next
    curr.next=new_node
    self.n+=1

  def insert_tail(self,val):

    tail_node=Node(val);
    if self.head==None:
      self.head=tail_node
      self.n+=1
      return

    curr = self.head
    while curr.next!=None:
      curr=curr.next
    curr.next=tail_node
    self.n+=1

  def clear(self):
    self.head=None
    self.n=0

  def delete_head(self):
    if self.head==None:
      return 'Empty Linked List'
    self.head=self.head.next
    self.n-=1

  def delete_tail(self):
    curr=self.head
    for i in range(self.n-2):
      curr=curr.next
    curr.next=None
    self.n-=1

  def delete_by_val(self,value):
    if self.head==None:
      return 'Empty Linked List'

    curr=self.head

    idx=0

    for i in range(self.n):
      if(curr.data == value):
        idx=i
      curr=curr.next

    curr=self.head

    if idx==0:
      self.head=curr.next
      self.n-=1
      return

    for _ in range(idx-1):
      curr=curr.next

    next_node=curr.next
    curr.next=next_node.next
    self.n-=1

  def find_by_val(self,value):
    curr=self.head
    for i in range(self.n):
      if (curr.data==value):
        return i
      curr=curr.next
    return "Value not in Linked list"

  def __getitem__(self,idx):
    curr=self.head
    pos=0
    while curr!=None:
      if idx==pos:
        return curr.data
      curr=curr.next
      pos+=1
    return 'Index Error'

  def __delitem__(self,idx):
    if self.head==None:
      return 'Empty Linked List'
    curr=self.head

    if idx==0:
      self.head=curr.next
      self.n-=1
      return

    for i in range(idx-1):
      curr=curr.next
    next=curr.next.next
    curr.next=next
    self.n-=1

  def __str__(self):
    curr=self.head
    res=''
    while curr!=None:
    # for i in range(self.n):
      res=res+str(curr.data)+'->'
      curr=curr.next
    return res[:-2]
```
