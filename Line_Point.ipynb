```python
class Point:
  def __init__(self,x,y):
    self.x=x
    self.y=y

  def __str__(self):
    return '<{},{}>'.format(self.x,self.y)

  def eucledian_distance(self,other):
    return ((other.x-self.x)**2+(other.y-self.y)**2)**0.5
  
  def distance_from_origin(self):
    return self.eucledian_distance(Point(0,0))

class Line:
  def __init__(self,A,B,C):
    self.A=A
    self.B=B
    self.C=C
  
  def __str__(self):
    return '{}x +{}y +{} =0'.format(self.A,self.B,self.C)
  
  def point_on_line(line,Point):
    if line.A*Point.x + line.B*Point.y + line.C == 0:
      return 'lies on the line'
    else:
      return 'does not lie on the line'
    
  def shortest_distance(line,point):
    return abs(line.A*point.x+line.B*point.y+line.C)/line.A**2+line.B**2
```

```python
p1=Point(0,0)
p2=Point(1,1)
p2.eucledian_distance(p1)
```
