```python
import ctypes
```

```python
class MyList():
  def __init__(self):
    self.n=0
    self.size=1
    self.A=self.__make_list(self.size)

  def __str__(self):
    stri=''
    for i in range(self.n):
      stri+=str(self.A[i])+','
    return '['+stri[:-1]+']'

  def __getitem__(self,index):
    if 0<=index<self.n:
      return self.A[index]
    else:
      return 'Index Error - Index out of bounds'

  def __delitem__(self,index):
    if 0<=index<self.n:
      for a in range(index,self.n-1):
        self.A[a]=self.A[a+1]
      self.n-=1

  def insert(self,pos,item):
    if self.n==self.size:
      self.__resize(self.size*2)

    for a in range(self.n,pos,-1):
      self.A[a]=self.A[a-1]

    self.A[pos]=item
    self.n+=1

  def append(self,item):
    if self.size == self.n:
      self.__resize(self.size*2)

    self.A[self.n]=item
    self.n+=1

  def find(self,item):
    for i in range(self.n):
      if self.A[i]==item:
        return i
    return 'Value - not in list'

  def remove(self,item):
    pos=self.find(item)

    if type(pos)==int:
      self.__delitem__(pos)
    else:
      return pos

  def clear(self):
    self.n=0
    self.size=1

  def pop(self):
    if self.n==0:
      return 'Empty List'
    print(self.A[self.n-1])
    self.n-=1

  def __resize(self,cap):
    B=self.__make_list(cap)
    self.size=cap

    for i in range(self.n):
      B[i]=self.A[i]
    self.A=B

  def __len__(self):
    return self.n

  def __make_list(self,capacity):
    return (capacity*ctypes.py_object)()
```

```python
l=MyList()
l.append(10)
l.append("hello")
l.append('jiji')
print(len(l))
print(l)
```
